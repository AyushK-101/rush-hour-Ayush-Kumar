# 01. Exploratory Data Analysis (EDA) - Project RUSH HOUR

This notebook explores the training data (`train_cells_capped.csv`) across the 6 maps.
- Class balance analysis (`free_flow`, `congested`, `blocked`)
- Feature distributions and correlation analysis
- Spatial visualization of grid patterns (60 x 60)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('../train_cells_capped.csv')
print('Dataset Shape:', df.shape)
print(df.head())
```

```python
# Overall Class Balance
print('--- Class Balance Across All Maps ---')
print(df['congestion_label'].value_counts(normalize=True))

# Class balance map by map
print('
--- Class Balance Map by Map ---')
print(pd.crosstab(df['map_id'], df['congestion_label'], normalize='index'))
```

```python
# Feature Summary by Congestion Class
feats = ['avg_speed_kmh', 'traffic_density_0_1', 'signal_delay_sec', 'intersection_density_0_1', 'incident_probability_0_1', 'simulated_travel_cost']
df.groupby('congestion_label')[feats].mean().T
```

```python
# Spatial Visualization: Visualizing Map 0 Grid
code_map = {'free_flow': 0, 'congested': 1, 'blocked': 2}
map0 = df[df['map_id'] == 0].sort_values(['row', 'col'])
grid0 = map0['congestion_label'].map(code_map).to_numpy().reshape(60, 60)

plt.figure(figsize=(7, 7))
plt.imshow(grid0, cmap='RdYlGn_r')
plt.title('Map 0 Spatial Layout (Green=Free, Orange=Congested, Red=Blocked)', fontsize=12)
plt.xlabel('Column')
plt.ylabel('Row')
plt.colorbar(ticks=[0, 1, 2], label='Congestion State')
plt.show()
```
